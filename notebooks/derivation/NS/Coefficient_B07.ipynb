## Corrected symbolic equivalence check

Optional clean Python/SymPy recomputation. This checks the ordinary and observer integrals against the existing exports; it does not validate their matrix-placement domain.

````python
# Corrected Python recomputation: explicit symbols, no Wolfram state.
import sympy
from IPython.display import display

t = sympy.Symbol('t', positive=True)
a = sympy.Symbol('a', positive=True)
b, c, p, z = sympy.symbols('b c p z', real=True)
# Ordinary interval: 0 < a < 1; all other parameters are real.
Q = 1-p*(1-t)/a
Z = 1-z*(1-t)/a
INTEGRAND = (b + c*t)*(1-t)/(a*t)*Q*Z
# Integrate an explicit Laurent polynomial; t and 1-a are positive.
ANTIDERIVATIVE = sympy.integrate(sympy.expand(INTEGRAND), t)
INTEGRAL = ANTIDERIVATIVE.subs(t, 1)-ANTIDERIVATIVE.subs(t, 1-a)
OBSERVER = sympy.integrate(sympy.expand((b+c*t)*(1-t)/t*t**3*(1-z*(1-t))), (t, 0, 1))
# Exact unchanged .txt export copied for a data-independent notebook.
EXPORT_TEXT = 'I7 = (a * ( - 12 * b * p * z + 2 * a ** 2 * b * ( - 6 + 3 * p + 3 * z - 2 * p * z) + 6 * a * b * (2 * p + 2 * z - p * z) + a ** 3 * c * (6 - 4 * p - 4 * z + 3 * p * z)) - 12 * b * (a - p) * (a - z) * numpy.log(1 - a)) / (12 * a ** 3)\n\nJ7 = (1 / 60) * (5 * b + 3 * c - (2 * b + c) * z)'
EXPORTS = {}
for LINE in EXPORT_TEXT.split('\n\n'):
    if '=' in LINE:
        NAME, EXPRESSION = LINE.split('=', 1)
        EXPORTS[NAME.strip()] = sympy.sympify(
            EXPRESSION.strip().replace('\n', ' ').replace('numpy.log', 'log').replace('^', '**'),
            locals={'a': a, **{symbol.name: symbol for symbol in INTEGRAND.free_symbols}, 'log': sympy.log},
        )
ORDINARY_RESIDUAL = sympy.simplify(INTEGRAL-EXPORTS['I7'])
OBSERVER_RESIDUAL = sympy.simplify(OBSERVER-EXPORTS['J7'])
assert ORDINARY_RESIDUAL == 0 and OBSERVER_RESIDUAL == 0
display(INTEGRAL, OBSERVER, ORDINARY_RESIDUAL, OBSERVER_RESIDUAL)
````
