$$
B^{i = n < N, j = n < N}_{\phi \kappa, n} = \frac{3 \Omega_{\mathrm{m}, 0} H_0^2}{2 c^2} \int_{\chi_{n}}^{\chi_{n + 1}} \mathrm{d} \chi \left( \frac{\chi_{n + 1}^2 - \chi^2}{2 \left( \chi_{n + 1} - \chi_{n} \right)} - \frac{\chi \chi_{n + 1}}{\left( \chi_{n + 1} - \chi_{n} \right)} \ln{\frac{\chi_{n + 1}}{\chi}} \right)\left(\frac{\chi_{n + 1} - \chi}{\chi (\chi_{n + 1} - \chi_n)}\right) \left( \frac{\chi_{n + 1} - \chi}{\chi_{n + 1} - \chi_n} P_\ell^{n} + \frac{\chi - \chi_n}{\chi_{n + 1} - \chi_n} P_\ell^{n + 1} \right) \left(1 + \frac{\chi_{n + 1} - \chi}{\chi_{n + 1} - \chi_n} z_{n} + \frac{\chi - \chi_{n}}{\chi_{n + 1} - \chi_{n}} z_{n + 1} \right)
$$

$$
B^{i = 0, j = 0}_{\phi \kappa, 0} = \frac{3 \Omega_{\mathrm{m}, 0} H_0^2}{2 c^2} \int_{0}^{\chi_{1}} \mathrm{d} \chi \left( \frac{\chi_{1}^2 - \chi^2}{2 \chi_{1} } - \chi \ln{\frac{\chi_{1}}{\chi}} \right)\left(\frac{\chi_{1} - \chi}{\chi \chi_{1} }\right) \left( \frac{\chi^3}{\chi_{1}^3}P_\ell^{1} \right) \left(1 +  \frac{\chi }{\chi_{1}} z_{1} \right)
$$

$$
z = \frac{z_{n + 1}- z_{n}}{1 + z_{n + 1}}

$$
a = \frac{\chi_{n + 1}- \chi_{n}}{\chi_{n + 1}}
$$

[Source cell 6](#source-06), `InlineFormula`.

$$
I_1 = \int_{1-a}^{1} \mathrm{d} t \: \left( \frac{1 - t^2}{2 a} + \frac{t \ln{t}}{a} \right)\left( \frac{1-t}{a t}\right)\left( 1 - \frac{p}{a} (1 - t) \right) \left( 1 - \frac{z}{a} (1 - t) \right)
$$

[Source cell 7](#source-07), `InlineFormula`.

$$
J_1 = \int_{0}^{1} \mathrm{d} t \: \left( \frac{1 - t^2}{2 } +t \ln{t}\right) \left( \frac{1-t}{t} \right) t^3  \left( 1 - z (1 - t) \right)
$$
