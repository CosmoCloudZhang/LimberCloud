$$
B^{i  = n+1 < N, n+1 < j = k  < N}_{\phi \kappa,n} = \frac{3 \Omega_{\mathrm{m}, 0} H_0^2}{2 c^2}  \int_{\chi_{n}}^{\chi_{n + 1}} \mathrm{d} \chi  \left(\frac{\chi_{k + 1}-\chi_{k - 1}}{2}-\frac{\chi \chi_{k + 1}}{\chi_{k + 1}-\chi_{k}} \ln{\frac{\chi_{k + 1}}{\chi_{k}}}+\frac{\chi \chi_{k - 1}}{\chi_{k}-\chi_{k - 1}} \ln{\frac{\chi_{k}}{\chi_{k - 1}}} \right)\left(\frac{\chi - \chi_n}{\chi (\chi_{n + 1} - \chi_n)}\right) \left( \frac{\chi_{n + 1} - \chi}{\chi_{n + 1} - \chi_n} P_\ell^{n} + \frac{\chi - \chi_n}{\chi_{n + 1} - \chi_n} P_\ell^{n + 1} \right) \left(1 + \frac{\chi_{n + 1} - \chi}{\chi_{n + 1} - \chi_n} z_{n} + \frac{\chi - \chi_{n}}{\chi_{n + 1} - \chi_{n}} z_{n + 1} \right)
$$

[Source cell 2](#source-02), `InlineFormula`.

$$
B^{i  = 1, 1 < j=k  < N}_{\phi \kappa, 0} = \frac{3 \Omega_{\mathrm{m}, 0} H_0^2}{2 c^2}  \int_{0}^{\chi_{1}} \mathrm{d} \chi  \left(\frac{\chi_{k + 1}-\chi_{k - 1}}{2}-\frac{\chi \chi_{k + 1}}{\chi_{k + 1}-\chi_{k}} \ln{\frac{\chi_{k + 1}}{\chi_{k}}}+\frac{\chi \chi_{k - 1}}{\chi_{k}-\chi_{k - 1}} \ln{\frac{\chi_{k}}{\chi_{k - 1}}} \right)\left(\frac{1}{\chi_{1} }\right) \left( \frac{\chi^3}{\chi_{1}^3}P_\ell^{1} \right) \left(1 + \frac{\chi }{\chi_{1} } z_{1} \right)
$$

[Source cell 3](#source-03), `InlineFormula`.

$$
t = \frac{\chi}{\chi_{n + 1}}
$$

$$
p = \frac{P_\ell ^{n + 1}- P_\ell^{n}}{P_\ell^{n + 1}}
$$

$$
z = \frac{z_{n + 1}- z_{n}}{1 + z_{n + 1}}
$$

$$
a = \frac{\chi_{n + 1}- \chi_{n}}{\chi_{n + 1}}
$$

$$
b = \frac{\chi_{k + 1}-\chi_{k - 1}}{2\chi_{n + 1}}
$$

$$
c=\frac{\chi_{k - 1}}{\chi_{k}-\chi_{k - 1}} \ln{\frac{\chi_{k}}{\chi_{k - 1}}}-\frac{\chi_{k + 1}}{\chi_{k +1}-\chi_{k}} \ln{\frac{\chi_{k+1}}{\chi_{k}}}
$$

$$
B^{i  = n+1, n+1 < j=k  < N}_{\phi \kappa, n}  = \frac{3 \Omega_{\mathrm{m}, 0} H_0^2}{2 c^2} \chi_{n + 1} P_\ell^{n + 1} (1 + z_{n + 1}) I_4
$$

[Source cell 5](#source-05), `InlineFormula`.

$$
B^{i  = 1, 1 < j=k  < N}_{\phi \kappa, 0}  = \frac{3 \Omega_{\mathrm{m}, 0} H_0^2}{2 c^2} \chi_{1} P_\ell^{1} (1 + z_{1}) J_4
$$

[Source cell 6](#source-06), `InlineFormula`.

$$
I_4 = \int_{1-a}^{1} \mathrm{d} t \:  \left(  b + c t \right)\left( \frac{1}{a}+\frac{a-1}{a t} \right) \left( 1 - \frac{p}{a} (1 - t) \right) \left( 1 - \frac{z}{a} (1 - t) \right)
$$

[Source cell 7](#source-07), `InlineFormula`.

$$
J_4 = \int_{0}^{1} \mathrm{d} t \:  \left(  b + c t \right) t^3 \left( 1 - z(1 - t) \right)
$$
