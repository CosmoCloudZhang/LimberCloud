$$
B^{i = n+1 < N, j = N}_{\phi \kappa,n}= \frac{3 \Omega_{\mathrm{m}, 0} H_ 0^2}{2 c^2} \int_{\chi_{n}}^{\chi_{n + 1}} \mathrm{d} \chi  
\left( -\chi+\frac{\chi_{N} -\chi_{N-1} }{2}+\frac{\chi \chi_{N-1} }{\chi_{N} -\chi_{N-1} } \ln{\frac{\chi_{N} }{\chi_{N-1} }} \right)
 \left(\frac{\chi - \chi_n}{\chi (\chi_{n + 1} - \chi_n)}\right) \left( \frac{\chi_{n + 1} - \chi}{\chi_{n + 1} - \chi_n} P_\ell^{n} + \frac{\chi - \chi_n}{\chi_{n + 1} - \chi_n} P_\ell^{n + 1} \right) \left(1 + \frac{\chi_{n + 1} - \chi}{\chi_{n + 1} - \chi_n} z_{n} + \frac{\chi - \chi_{n}}{\chi_{n + 1} - \chi_{n}} z_{n + 1} \right)
$$

B^{i = 1, j = N}_{\phi \kappa, 0}= \frac{3 \Omega_{\mathrm{m}, 0} H_ 0^2}{2 c^2} \int_{0}^{\chi_{1}} \mathrm{d} \chi  
\left( -\chi+\frac{\chi_{N} -\chi_{N-1} }{2}+\frac{\chi \chi_{N-1} }{\chi_{N} -\chi_{N-1} } \ln{\frac{\chi_{N} }{\chi_{N-1} }} \right)
