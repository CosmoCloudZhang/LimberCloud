B^{i = n+1 < N, j = n+1 < N}_{\phi \kappa,n} = \frac{3 \Omega_{\mathrm{m}, 0} H_0^2}{2 c^2} \int_{\chi_{n}}^{\chi_{n + 1}} \mathrm{d} \chi \left( \frac{\chi_{n + 2}-\chi_{n + 1}}{2}-\frac{\chi \chi_{n + 2}}{\chi_{n + 2}-\chi_{n + 1}}\ln{\frac{\chi_{n + 2}}{\chi_{n + 1}}}+\frac{\chi_{n + 1}^2-2\chi_{n}\chi_{n + 1}+\chi^2}{2\left( \chi_{n + 1}-\chi_{n}  \right)}+\frac{\chi \chi_{n}}{\chi_{n + 1}-\chi_{n}}\ln{\frac{\chi_{n + 1}}{\chi}} \right) \left(\frac{\chi - \chi_n}{\chi (\chi_{n + 1} - \chi_n)}\right) \left( \frac{\chi_{n + 1} - \chi}{\chi_{n + 1} - \chi_n} P_\ell^{n} + \frac{\chi - \chi_n}{\chi_{n + 1} - \chi_n} P_\ell^{n + 1} \right) \left(1 + \frac{\chi_{n + 1} - \chi}{\chi_{n + 1} - \chi_n} z_{n} + \frac{\chi - \chi_{n}}{\chi_{n + 1} - \chi_{n}} z_{n + 1} \right)
$$

$$
