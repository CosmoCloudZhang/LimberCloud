$$
B^{i = n+1 < N, j = n+1 < N}_{\phi \phi, n} = \int_{\chi_{n}}^{\chi_{n + 1}} \mathrm{d} \chi \left(\frac{\chi - \chi_n}{\chi (\chi_{n + 1} - \chi_n)}\right)^2 \left( \frac{\chi_{n + 1} - \chi}{\chi_{n + 1} - \chi_n} P_\ell^{n} + \frac{\chi - \chi_n}{\chi_{n + 1} - \chi_n} P_\ell^{n + 1} \right)
$$

[Source cell 2](#source-02), `InlineFormula`.

$$
B^{i = 1, j = 1}_{\phi \phi, 0} = \int_{0}^{\chi_{1}} \mathrm{d} \chi \left(\frac{1}{\chi_{1}}\right)^2 \left( \frac{\chi^3}{\chi_{1}^3}P_\ell^{1} \right)
$$

[Source cell 3](#source-03), `InlineFormula`.

$$
t = \frac{\chi}{\chi_{n + 1}}
$$

$$
p = \frac{P_\ell ^{n + 1}- P_\ell^{n}}{P_\ell^{n + 1}}
$$

$$
a = \frac{\chi_{n + 1}- \chi_{n}}{\chi_{n + 1}}
$$

[Source cell 4](#source-04), `InlineFormula`.

$$
B^{i = n+1 < N, j = n+1 < N}_{\phi \phi, n} =\frac{P_\ell^{n + 1}}{\chi_{n+1}}  I_3

B^{i = 1, j =1}_{\phi \phi, 0} =\frac{P_\ell^{1}}{\chi_1}  J_3

$$
I_3 = \int_{1-a}^{1} \mathrm{d} t \:  \left( \frac{1}{a}+\frac{a-1}{a t} \right)^2 \left( 1 - \frac{p}{a} (1 - t) \right)
$$

[Source cell 7](#source-07), `InlineFormula`.

$$
J_3 = \int_{0}^{1}   t^3 \mathrm{d} t
$$
