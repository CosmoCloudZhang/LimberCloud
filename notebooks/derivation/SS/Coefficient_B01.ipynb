$$
B^{i = n < N, j = n < N}_{\kappa \kappa, n}= \frac{9 \Omega_{\mathrm{m}, 0}^2 H_0^4}{4 c^4} \int_{\chi_{n}}^{\chi_{n + 1}} \mathrm{d} \chi \left( \frac{\chi_{n + 1}^2 - \chi^2}{2 \left( \chi_{n + 1} - \chi_{n} \right)} - \frac{\chi \chi_{n + 1}}{\left( \chi_{n + 1} - \chi_{n} \right)} \ln{\frac{\chi_{n + 1}}{\chi}} \right)^2 \left( \frac{\chi_{n + 1} - \chi}{\chi_{n + 1} - \chi_n} P_\ell^{n} + \frac{\chi - \chi_n}{\chi_{n + 1} - \chi_n} P_\ell^{n + 1} \right) \left(1 + \frac{\chi_{n + 1} - \chi}{\chi_{n + 1} - \chi_n} z_{n} + \frac{\chi - \chi_{n}}{\chi_{n + 1} - \chi_{n}} z_{n + 1} \right)^2
$$

$$
B^{i = 0, j = 0}_{\kappa \kappa, 0}= \frac{9 \Omega_{\mathrm{m}, 0}^2 H_0^4}{4 c^4} \int_{0}^{\chi_{1}} \mathrm{d} \chi \left( \frac{\chi_{1}^2 - \chi^2}{2 \chi_{1} } - \chi  \ln{\frac{\chi_{1}}{\chi}} \right)^2 \left( \frac{\chi^3}{\chi_{1}^3}P_\ell^{1} \right) \left(1 + \frac{\chi}{\chi_{1} } z_{1} \right)^2
$$

[Source cell 3](#source-03), `InlineFormula`.

$$
t = \frac{\chi}{\chi_{n + 1}}
$$

$$
p = \frac{P_\ell ^{n + 1}- P_\ell^{n}}{P_\ell^{n + 1}}
$$

$$
z = \frac{z_{n + 1}- z_{n}}{1 + z_{n + 1}}
$$

$$
a = \frac{\chi_{n + 1}- \chi_{n}}{\chi_{n + 1}}
$$

[Source cell 6](#source-06), `InlineFormula`.

$$
I_1 = \int_{1-a}^{1} \mathrm{d} t \: \left( \frac{1 - t^2}{2 a} + \frac{t \ln{t}}{a} \right)^2 \left( 1 - \frac{p}{a} (1 - t) \right) \left( 1 - \frac{z}{a} (1 - t) \right)^2
$$

[Source cell 7](#source-07), `InlineFormula`.

$$
J_1 = \int_{0}^{1} \mathrm{d} t \: \left( \frac{1 - t^2}{2} + x \ln{t} \right)^2 t^3 \left( 1 - z(1 - t) \right)^2
$$
