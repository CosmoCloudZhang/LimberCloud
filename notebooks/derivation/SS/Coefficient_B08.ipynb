$$
B^{n+1 < i  < N, n+1 < j  < N}_{\kappa \kappa,n}= \frac{9 \Omega_{\mathrm{m}, 0}^2 H_ 0^4}{4 c^4} \int_{\chi_{n}}^{\chi_{n + 1}} \mathrm{d} \chi \left(\frac{\chi_{i + 1}-\chi_{i - 1}}{2}-\frac{\chi \chi_{i + 1}}{\chi_{i + 1}-\chi_{i}} \ln{\frac{\chi_{i + 1}}{\chi_{i}}}+\frac{\chi \chi_{i - 1}}{\chi_{i}-\chi_{i - 1}} \ln{\frac{\chi_{i}}{\chi_{i - 1}}} \right)  \left(\frac{\chi_{j + 1}-\chi_{j - 1}}{2}-\frac{\chi \chi_{j + 1}}{\chi_{j + 1}-\chi_{j}} \ln{\frac{\chi_{j + 1}}{\chi_{j}}}+\frac{\chi \chi_{j - 1}}{\chi_{j}-\chi_{j - 1}} \ln{\frac{\chi_{j}}{\chi_{j - 1}}} \right) \left( \frac{\chi_{n + 1} - \chi}{\chi_{n + 1} - \chi_n} P_\ell^{n} + \frac{\chi - \chi_n}{\chi_{n + 1} - \chi_n} P_\ell^{n + 1} \right) \left(1 + \frac{\chi_{n + 1} - \chi}{\chi_{n + 1} - \chi_n} z_{n} + \frac{\chi - \chi_{n}}{\chi_{n + 1} - \chi_{n}} z_{n + 1} \right)^2
$$

$$
